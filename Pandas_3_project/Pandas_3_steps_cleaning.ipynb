```python
import warnings
warnings.filterwarnings('ignore')

```

```python

df['Cuisines'].fillna('Unknown', inplace=True)
```

```python
df.isnull().sum()
```

```python
bool_cols = ['Has Table booking', 'Has Online delivery', 'Is delivering now', 'Switch to order menu']

for col in bool_cols:
    df[col] = df[col].apply(lambda x: True if str(x).strip().lower() == 'yes' else False)
```

```python
df.head()
```

```python
country_map = {
    1: 'India', 14: 'Australia', 30: 'Brazil', 37: 'Canada',
    94: 'Indonesia', 148: 'New Zealand', 162: 'Philippines',
    166: 'Qatar', 184: 'Singapore', 189: 'South Africa',
    191: 'Sri Lanka', 208: 'Turkey', 214: 'UAE',
    215: 'United Kingdom', 216: 'United States'
}

df['Country'] = df['Country Code'].map(country_map)
```

```python
df.head()
```

```python
df.duplicated().sum()
df.drop_duplicates(inplace=True)
```

```python
df['Average Cost for two'] = pd.to_numeric(df['Average Cost for two'], errors='coerce')
df['Price range']=pd.to_numeric(df['Price range'], errors='coerce')
```

```python
df.dtypes
```

```python
df.head()
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_').str.replace('(', '').str.replace(')', '')
```

```python
df.head()
```

```python
df.to_csv("cleaned_zomato.csv", index=False)
```
