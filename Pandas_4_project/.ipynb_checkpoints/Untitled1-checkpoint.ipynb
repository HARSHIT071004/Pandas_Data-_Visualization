```python
import pandas as pd
import matplotlib.pyplot as plt
from wordcloud import WordCloud

df=pd.read_csv("cleaned_amazon_reviews.csv")
df.head()
```

```python
top_rated= df.groupby('product_name')['rating_count'].sum().sort_values(ascending=False).head(10)
top_rated.plot(kind='barh', figsize=(10,6), color='skyblue')
plt.title('Top 10 Most Rated Products')
plt.xlabel('Total Ratings')
plt.ylabel('Product Name')
plt.tight_layout()
plt.show() 
```

```python
avg_rating = df.groupby('category')['rating'].mean().sort_values(ascending=False).head(10)

avg_rating.plot(kind='barh', figsize=(10,5), color='orange')
plt.title('Top 10 Categories by Average Rating')
plt.xlabel('Average Rating')
plt.ylabel('Category')
plt.tight_layout()
plt.show()


```

```python
df['rating'].dropna().hist(bins=10, color='purple', edgecolor='black', figsize=(8,4))
plt.title('Distribution of Product Ratings')
plt.xlabel('Rating')
plt.ylabel('Number of Products')
plt.tight_layout()
plt.show()
```

```python
text = ' '.join(df['review_content'].dropna().astype(str).tolist())

wordcloud = WordCloud(width=1000, height=500, background_color='white').generate(text)

plt.figure(figsize=(15,7))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title('Most Common Words in Reviews')
plt.tight_layout()
plt.show()
```

```python
top_discount = df.sort_values('discount_percentage', ascending=False).head(20)

plt.scatter(top_discount['discount_percentage'], top_discount['rating'], color='green', s=100)
plt.title('Top 10 Highest Discount vs Rating')
plt.xlabel('Discount Percentage')
plt.ylabel('Rating')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python

top_10 = df.groupby('product_name')['rating_count'].sum().sort_values(ascending=False).head(10).index
top_10_df = df[df['product_name'].isin(top_10)]
plt.figure(figsize=(10, 5))
plt.scatter(top_10_df['actual_price'], top_10_df['rating'], alpha=0.6, color='red')
plt.xscale('log')  # Optional: shows better distribution if price range is large
plt.title('Price vs Rating (Top 10 Most Rated Products)')
plt.xlabel('Actual Price (log scale)')
plt.ylabel('Rating')
plt.tight_layout()
plt.show()
```

```python
df.to_csv('Visualized_amazon_reviews.csv', index=False)
```
