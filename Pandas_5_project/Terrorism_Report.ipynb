```python
# Importing Necessary Libararies
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
# Loading Datasets
df = pd.read_csv('C:\\Users\\nikrc\\OneDrive\\Desktop\\Datasets\\CSV\\terrorism.csv',encoding = 'ISO-8859-1',index_col = 'eventid')
pd.set_option('display.max_columns',None)
pd.set_option('display.max_rows',None)
df.head(5)
```

```python
# Checking Shape
print(f"The Number of rows are {df.shape[0]} and columns are: {df.shape[1]}")
```

```python
# Checking Null Values
print(df.isna().mean().T * 100)
```

```python
# Choosing Columns
# print(df['extended'].unique())
df.rename(columns = {"success":"WasSuccessful","iyear": "Year","imonth":"Month","iday":"Day","country_txt":"Country","region_txt":"Region","city":"City","latitude":"Latitude","longitude":"Longitude","attacktype1_txt":"AttackType","targtype1_txt":"TargetType","targsubtype1_txt":"TargetSubType","corp1":"Corporation","target1":"Targeted Object","weaptype1_txt":"WeaponUsed","nkill":"Kills",'nwound':'WoundedCount',"gname":"GroupName"},inplace = True)
revised_df = df[["Year","Month","Day","Country","Region","City","Latitude","Longitude","AttackType","TargetType","TargetSubType","Corporation","Targeted Object","WeaponUsed","WasSuccessful","WoundedCount","Kills","GroupName"]].copy(deep = True)
revised_df.head(10)
```

```python
# Checking Null Values
print(revised_df.isna().mean() * 100)
```

```python
# Fixing NaN Values in TargetSubType 
revised_df['TargetSubType'] = revised_df.apply(lambda x: x['TargetType'] if pd.isna(x['TargetSubType']) else x['TargetSubType'],axis = 1)
revised_df['Corporation'] = revised_df.apply(lambda x: x['TargetSubType'] if pd.isna(x['Corporation']) else x['Corporation'],axis = 1)
revised_df.head(10)
```

```python
# ReChecking Null Values

print(revised_df.isna().mean() * 100)
```

```python
# Clearing Out Kills and WoundedCount
revised_df['Kills'].fillna(df['Kills'].median(),inplace=True)
revised_df['WoundedCount'].fillna(df['WoundedCount'].median(),inplace=True)
revised_df.dropna(subset=['City','Latitude','Longitude','Targeted Object'],inplace=True)
```

```python
# Dropping Duplicates
revised_df= revised_df.drop_duplicates(keep = False)
```

```python
# Feature Engineering
revised_df['Casualities'] = revised_df['Kills'] + revised_df['WoundedCount']
```

```python
# Checking Data Types
revised_df.info()
```

```python
# 1: Univariate Analysis
# 2: Bivariate Analysis
```

```python
# Number of Terrorist Activitues Each Year
plt.figure(figsize = (15,6))
sns.countplot(x = 'Year',data = revised_df,palette='vlag',edgecolor = sns.color_palette('dark',7))
plt.xticks(rotation = 90)
plt.title('Number of Terrorist Activitues Each Year')
plt.show()
```

```python
# Conclusion
# Based on the above plot, We can observe that since 1970's the attacks have been on rise for around One and half a decade
# and due to some factors attacks declined for 7 years, and again on rise from 2005 till Now.
```

```python
# Number of Terrorist Attaks in Each Country
plt.figure(figsize = (15,6))
top_10_countries = revised_df['Country'].value_counts()[:10].to_frame().reset_index()
sns.barplot(x = 'Country',y = 'count',data = top_10_countries,palette='mako',edgecolor = sns.color_palette('dark',7))
plt.xticks(rotation = 90)
plt.title('Number of Terrorist Attaks in Each Country')
plt.show()
```

```python
# Number of Terrorist Attaks by Regions
plt.figure(figsize = (15,6))
top_10_countries = revised_df['Region'].value_counts()[:7].to_frame().reset_index()
sns.barplot(x = 'Region',y = 'count',data = top_10_countries,palette='mako',edgecolor = sns.color_palette('dark',7))
plt.xticks(rotation = 90)
plt.title('Number of Terrorist Attaks by Regions')
plt.show()
# pd.crosstab(revised_df.Year,revised_df.Region).plot(kind = 'area',figsize = (15,6),colormap = 'vlag')
```

```python
# Most Weapons Used by Terrorists
plt.figure(figsize = (15,6))
top_5_weapons = revised_df['WeaponUsed'].value_counts()[:5].to_frame().reset_index().sort_values(by='count',ascending = True)
sns.barplot(x = 'WeaponUsed',y = 'count',data = top_5_weapons,palette='vlag',edgecolor = sns.color_palette('dark',7))
plt.xticks(rotation = 90)
plt.title('Most Weapons Used by Terrorists')
plt.show()
# pd.crosstab(revised_df.Year,revised_df.Region).plot(kind = 'area',figsize = (15,6),colormap = 'vlag')
```

```python
# Number Attacks by a Terrorist Group
plt.figure(figsize = (15,6))
top_5_group = revised_df['GroupName'].value_counts()[:5].to_frame().reset_index().sort_values(by='count',ascending = True)
sns.barplot(x = 'GroupName',y = 'count',data = top_5_group,palette='vlag',edgecolor = sns.color_palette('dark',7))
plt.xticks(rotation = 90)
plt.title('Number Attacks by a Terrorist Group')
plt.show()
```

```python
# How much each weapon category being used by terrorists 

top_5_weapons = revised_df['WeaponUsed'].value_counts()[:5]
top_5_weapons.plot(kind = 'pie',autopct = '%1.1f%%',figsize = (8,6),colormap='vlag')
plt.show()
```

```python
# Number of Total Casualties in Each Country & Year

fig,axes = plt.subplots(2,1,figsize = (20,12),dpi = 120)
required_cols = ['Region','Year']
for i,col in enumerate(required_cols):
    sns.barplot(x = col,y = 'Casualities',data = revised_df,palette='viridis',ax = axes[i])
    axes[i].tick_params(rotation = 90)
    axes[i].set_title(f"Number of Total Casualties by {'Region' if i == 0 else 'Year'}",fontweight = 'semibold')
    

plt.subplots_adjust(hspace=0.6)
plt.tight_layout()
plt.show()
```

```python
# Multivariate Analysis

# Number of Terrorist Attacks in East Asia,Middle East & North Africa Countries
Country_Wise_df = (
    revised_df[revised_df['Region'].isin(["East Asia", "Middle East & North Africa"])]['Country'].value_counts()[:10]
    .reset_index(name='count')
    .sort_values(by='count', ascending=False)
)
Year_Wise_df = (
    revised_df[revised_df['Region'].isin(["East Asia", "Middle East & North Africa"])]['Year'].value_counts()
    .reset_index(name='count')
    .sort_values(by='count', ascending=False)
)
# Country_Wise_df.head()
City_Wise_df = (
    revised_df[revised_df['Region'].isin(["East Asia", "Middle East & North Africa"])]['City'].value_counts()[:5]
    .reset_index(name='count')
    .sort_values(by='count', ascending=False)
)
# Country_Wise_df.head()
```

```python
plt.figure(figsize = (15,8))
sns.barplot(x = 'Country',y = 'count',data = Country_Wise_df,palette='viridis')
plt.xticks(rotation = 90)
plt.title("Number of Terrorist Attacks in East Asia,Middle East & North Africa Countries")
plt.show()
```

```python
plt.figure(figsize = (15,8))
sns.barplot(x = 'Year',y = 'count',data = Year_Wise_df,palette='viridis')
plt.xticks(rotation = 90)
plt.title("Most Active Terror Year in East Asia,Middle East & North Africa Countries")
plt.show()
```

```python
plt.figure(figsize = (15,8))
sns.barplot(x = 'count',y = 'City',data = City_Wise_df,palette='viridis')
plt.xticks(rotation = 90)
plt.title("Most Active Terror Year in East Asia,Middle East & North Africa Countries")
plt.show()
```

```python
# As we Know that terrorism have been on rise after 2007, We could check who have been the usual targets for these successful terrorist attacks in the most attacked country

targeted_insights = revised_df[(revised_df['Year'] > 2007) & (revised_df['WasSuccessful'] == 1)].groupby(by='Country').size().to_frame().reset_index()
targeted_insights.rename({0:'Attacks'},inplace=True,axis=1)
country_with_most_attacks = targeted_insights.loc[targeted_insights['Attacks'].idxmax()]
country = country_with_most_attacks['Country']
# targeted_insights.sort_values(by = 'Attacks',ascending=False).head()
targeted_data = revised_df[(revised_df['Year'] > 2007) & (revised_df['WasSuccessful'] == 1) & (revised_df['Country'] == country)]

targeted_data.head()
```

```python

plt.figure(figsize=(15,8))
sns.countplot(x = 'TargetType',data = targeted_data,palette='viridis')
plt.xticks(rotation = 90)
plt.show()
```
