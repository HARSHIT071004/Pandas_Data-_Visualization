```python
import pandas as pd
```

```python
df=pd.read_csv("IPL_2023_DATASET.csv")
```

```python
df.drop(columns=["Unnamed: 0"], inplace=True)
```

```python
print(df.columns)
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

for col in df.select_dtypes(include="object"):
    df[col] = df[col].str.strip()

```

```python
df.isnull().sum()
```

```python
df["base_price"].fillna("Unknown", inplace=True)

# Step 1: NaN को Unsold
df["cost_in_₹_(cr.)"].fillna("Unsold", inplace=True)

# Step 2: 0.0 को भी Unsold
df["cost_in_₹_(cr.)"] = df["cost_in_₹_(cr.)"].replace(0.0, "Unsold")

# Similarly for cost_in_$_(000)
df["cost_in_$_(000)"].fillna("Unsold", inplace=True)
df["cost_in_$_(000)"] = df["cost_in_$_(000)"].replace(0.0, "Unsold")

# 2022_squad null ko replace
df["2022_squad"].fillna("Not in Squad", inplace=True)

# df["Base Price IN ₹"].fillna(0, inplace=True)
# df["Base Price IN $"].fillna(0, inplace=True)
```

```python
# Base price in ₹
df["base_price_in_₹"] = df["base_price_in_₹"].fillna(0)
df["base_price_status"] = df["base_price_in_₹"].apply(lambda x: "Not Declared" if x == 0 else "Declared")

# Base price in $
df["base_price_in_$"] = df["base_price_in_$"].fillna(0)
df["base_price_status_$"] = df["base_price_in_$"].apply(lambda x: "Not Declared" if x == 0 else "Declared")

```

```python
df.head(500)
```

```python
df.dtypes
```

```python
df["cost_in_₹_(cr.)"] = (
    df["cost_in_₹_(cr.)"]
    .astype(str)
    .str.replace("₹", "", regex=False)
    .str.replace("Cr.", "", regex=False)
    .str.replace("Cr", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

# Step 2: "Unsold", "-", "" जैसी values को NaN बना दो
df["cost_in_₹_(cr.)"] = df["cost_in_₹_(cr.)"].replace(["Unsold", "-", ""], pd.NA)

# Step 3: अब float में convert करो
df["cost_in_₹_(cr.)"] = pd.to_numeric(df["cost_in_₹_(cr.)"], errors="coerce")

df["cost_in_$_(000)"] = df["cost_in_$_(000)"].replace(["Unsold", "-", ""], pd.NA)

df["cost_in_$_(000)"] = pd.to_numeric(df["cost_in_$_(000)"], errors="coerce")
# # Step 4 (optional): Null values को drop कर दो (ताकि plot साफ बने)
# # df = df.dropna(subset=["cost_in_₹_(cr.)"])

# # df["base_price_in_₹"] = df["base_price_in_₹"].astype(float)
# # df["base_price_in_$"] = df["base_price_in_$"].astype(float)
# # df["cost_in_₹_(cr.)"] = df["cost_in_₹_(cr.)"].astype(float)
# # df["cost_in_$_(000)"] = df["cost_in_$_(000)"].astype(float)

# # df.dtypes
```

```python
df["base_price_in_₹"] = df["base_price_in_₹"].astype(float)
# df["base_price_in_$"] = df["base_price_in_$"].astype(float)
df["cost_in_₹_(cr.)"] = df["cost_in_₹_(cr.)"].astype(float)
df["cost_in_$_(000)"] = df["cost_in_$_(000)"].astype(float)

df.dtypes
```
