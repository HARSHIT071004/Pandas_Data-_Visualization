# 🏏 IPL 2023 Data Analysis

This notebook presents a visual analysis of the IPL 2023 players dataset. We'll explore which teams spent the most, which players were the most expensive, and extract some key insights from the data.

## 🎯 Objective

- Analyze team-wise and player-wise spending.
- Visualize the top players and team composition.
- Identify cost patterns and potential trends in the dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from wordcloud import WordCloud
sns.set(style='whitegrid')
```

## 📥 Loading Dataset

```python
df = pd.read_csv("IPL_2023_CLEANED.csv")
df.head()
```

## 📊 Dataset Overview

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

## 💰 Top 10 Most Expensive Players

```python
top_10 = df.sort_values(by="cost_in_₹_(cr.)", ascending=False).head(10)
plt.figure(figsize=(8,6))
sns.barplot(data=top_10, x="player", y="cost_in_₹_(cr.)", hue="team")
plt.xticks(rotation=45)
plt.title("Top 10 Most Expensive Players")
plt.tight_layout()
plt.show()
```

## 🏢 Team-wise Total Spending

```python
team_spending = df.groupby("team")["cost_in_₹_(cr.)"].sum().sort_values(ascending=False)
plt.figure(figsize=(12,6))
team_spending.plot(kind="bar", color="skyblue")
plt.ylabel("Total Spending (₹ Cr.)")
plt.title("Team-wise Total Spending")
plt.xticks(rotation=45)
plt.show()
```

## ☁️ Word Cloud of Player Names

```python
text = " ".join(df["player"])
wordcloud = WordCloud(width=800, height=400, background_color='white').generate(text)
plt.figure(figsize=(10,5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title("Word Cloud of IPL 2023 Players")
plt.show()
```

## 📌 Key Insights

- The most expensive player in IPL 2023 was shown in the first plot.
- Some teams spent significantly more on their squad than others.
- Word cloud reveals frequent player names or popular selections.
- Further insights can be extracted by adding performance-based metrics.
